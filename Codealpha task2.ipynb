```python
# ==============================
# CODEALPHA TASK 2 - EDA
# ==============================

# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset

df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')


# --------------------------------
# 1. Display Dataset
# --------------------------------
print("First 5 Rows of Dataset")
display(df.head())

# --------------------------------
# 2. Meaningful Questions
# --------------------------------
print("\nMeaningful Questions Before Analysis:")
print("1. What factors affected survival?")
print("2. Did gender impact survival?")
print("3. Which passenger class had highest survival?")
print("4. Are there missing values?")
print("5. Are there outliers in the dataset?")

# --------------------------------
# 3. Explore Data Structure
# --------------------------------
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

# --------------------------------
# 4. Statistical Summary
# --------------------------------
print("\nStatistical Summary:")
display(df.describe())

# --------------------------------
# 5. Missing Values
# --------------------------------
print("\nMissing Values:")
print(df.isnull().sum())

# --------------------------------
# 6. Handle Missing Values
# --------------------------------
df['Age'].fillna(df['Age'].mean(), inplace=True)
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)

# Drop Cabin column
df.drop('Cabin', axis=1, inplace=True)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

# --------------------------------
# 7. Duplicate Values
# --------------------------------
print("\nDuplicate Rows:")
print(df.duplicated().sum())

# --------------------------------
# 8. Survival Count Plot
# --------------------------------
plt.figure(figsize=(6,4))
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.show()

# --------------------------------
# 9. Gender vs Survival
# --------------------------------
plt.figure(figsize=(6,4))
sns.countplot(x='Sex', hue='Survived', data=df)
plt.title("Gender vs Survival")
plt.show()

# --------------------------------
# 10. Passenger Class vs Survival
# --------------------------------
plt.figure(figsize=(6,4))
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.title("Passenger Class vs Survival")
plt.show()

# --------------------------------
# 11. Age Distribution
# --------------------------------
plt.figure(figsize=(8,5))
sns.histplot(df['Age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.show()

# --------------------------------
# 12. Correlation Heatmap
# --------------------------------
df_numeric = df.copy()

df_numeric['Sex'] = df_numeric['Sex'].map({
    'male':0,
    'female':1
})

numeric_cols = df_numeric.select_dtypes(include=np.number)

plt.figure(figsize=(10,6))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()

# --------------------------------
# 13. Boxplot for Outliers
# --------------------------------
plt.figure(figsize=(6,4))
sns.boxplot(x=df['Fare'])
plt.title("Fare Boxplot")
plt.show()

# --------------------------------
# 14. Trends and Patterns
# --------------------------------
print("\nTrends and Patterns:")
print("1. Females survived more than males.")
print("2. First class passengers had higher survival.")
print("3. Outliers detected in Fare column.")
print("4. Age slightly affected survival.")

# --------------------------------
# 15. Hypothesis Testing
# --------------------------------
print("\nHypothesis Testing:")
print("H0: Gender does not affect survival.")
print("H1: Gender affects survival.")

survival_gender = pd.crosstab(df['Sex'], df['Survived'])

print("\nSurvival Table:")
display(survival_gender)

# --------------------------------
# 16. Final Conclusion
# --------------------------------
print("\nConclusion:")
print("EDA successfully identified trends, patterns, and anomalies in Titanic dataset.")
print("Gender and passenger class strongly influenced survival.")
```
